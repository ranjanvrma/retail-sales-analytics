## Business Analysis

```python
# How did total sales change year-over-year from 2015 to 2018, and which year experienced the highest growth?
Sales_by_Year = df.groupby(df['Order Date'].dt.year)['Sales'].sum()
Percentage_Growth = Sales_by_Year.pct_change() * 100

result = pd.DataFrame({
    'Sales': Sales_by_Year,
    'Percentage Growth': Percentage_Growth
})

result = result.sort_values('Percentage Growth', ascending=False)
result
```

```python
# Which product category generates the most sales, and what percentage of total sales does it contribute?
Category_Sales = df.groupby('Category')["Sales"].sum()
Sales_Contribution = Category_Sales / Category_Sales.sum() * 100

result = pd.DataFrame({
    'Total Sales': Category_Sales,
    'Sales Contribution': Sales_Contribution
})

result = result.sort_values('Total Sales', ascending=False)
result
```

```python
# Which region generates the most sales, and how does each region contribute to overall sales?
Region_Sales = df.groupby('Region')['Sales'].sum()
Sales_Contribution = Region_Sales / Region_Sales.sum() * 100

result = pd.DataFrame({
    'Total Sales': Region_Sales,
    'Sales Contribution': Sales_Contribution
})

result = result.sort_values('Total Sales', ascending=False)
result
```

```python
# Who are the top 10 customers by sales, and what percentage of total sales do they collectively contribute?
Customer_Sales = df.groupby(['Customer ID', 'Customer Name'])["Sales"].sum()
Sales_Contribution = Customer_Sales / Customer_Sales.sum() * 100

result = pd.DataFrame({
    'Total Sales': Customer_Sales,
    'Sales Contribution': Sales_Contribution
})

result = result.nlargest(10, 'Total Sales')
result
```

```python
# Which 10 products generate the highest sales, and which product contributes the most?
Product_Sales = df.groupby('Product Name')['Sales'].sum()

result = pd.DataFrame({
    'Total Sales': Product_Sales,
})

result = result.nlargest(10, 'Total Sales')
result
```

```python
  # What percentage of customers have placed more than one order?
Customer_Orders = df.groupby('Customer ID')['Order ID'].nunique()
Multiple_Order_Customers = (Customer_Orders > 1).sum()
Percentage_Multiple_Orders = (Multiple_Order_Customers / Customer_Orders.count()) * 100

print(f"Percentage of Customers with More than One Order: {Percentage_Multiple_Orders:.2f}%")
```

```python
# Which shipping mode has the highest average shipping time, and how many unique orders use each shipping mode?
df['Shipping Time'] = df['Ship Date'] - df['Order Date']
Shipping_Time_by_Mode = df.groupby('Ship Mode')['Shipping Time'].mean()
Unique_Orders_by_Mode = df.groupby('Ship Mode')['Order ID'].nunique()

result = pd.DataFrame({
    'Average Shipping Time': Shipping_Time_by_Mode,
    'Unique Orders': Unique_Orders_by_Mode
})

result = result.sort_values('Average Shipping Time', ascending=False)
result
```

```python
# For each region, which product category generates the highest sales?
Region_Category_Sales = df.groupby(['Region', 'Category'])['Sales'].sum()
Top_idx = Region_Category_Sales.groupby(level = 'Region').idxmax()

result = Region_Category_Sales.loc[Top_idx].reset_index()
result.columns = ['Region', 'Top Category', 'Total Sales']
result
```

```python
# What percentage of total sales comes from the top 5% highest-value sales transactions?
Total_Sales = df['Sales'].sum()
Top_5_Percent_Sales = df.nlargest(int(0.05 * len(df)), 'Sales')['Sales'].sum()
Percentage_Top_5_Percent = (Top_5_Percent_Sales / Total_Sales) * 100

print(f"Percentage of Total Sales from Top 5% Highest-Value Transactions: {Percentage_Top_5_Percent:.2f}%")
```

```python
# Which sub-categories experienced the strongest sales growth between 2015 and 2018?
Sub_Category_Sales_2015 = df[df['Order Date'].dt.year == 2015].groupby('Sub-Category')['Sales'].sum()
Sub_Category_Sales_2018 = df[df['Order Date'].dt.year == 2018].groupby('Sub-Category')['Sales'].sum()
Sales_Growth = ((Sub_Category_Sales_2018 - Sub_Category_Sales_2015) / Sub_Category_Sales_2015) * 100

result = pd.DataFrame({
    'Strongest Sales Growth': Sales_Growth
}).sort_values(by='Strongest Sales Growth', ascending=False)

result
```
